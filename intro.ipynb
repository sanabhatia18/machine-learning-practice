MACHINE LEARNING : machine learns from past data and ables to predict unseen data 


~ ***Training data*** : by using training data, the machine learns patterns and relationships within the data to make accurate predictions on new, unseen data. we create ml model using it

~ ***Testing data*** : we evaluate how the model will perform on unseen data .

***RULE BASED PROGRAMMING VS MACHINE LEARNING*** :

~in rule based: we used conditions to get true or false results 

               : uses pre defined conditions to get results 
               
               : created by programmers

               : requires less data

               : depends on condition 


~ in ml : machine learns patterns from past data and ables to do prediction of unseen data.

                : we use models for prediction 

                : already created by data scientists known as algoriithms

                : requirement of data is very high

                : depends on pattern of past data
          

***TYPES OF MACHINE LEARNING*** : 
1. Supervised learning : ml which has labeled data.

   ~ independent data (x) : input data which is used to predict the output data. 
                            example: car price prediction : x= car features like mileage, year, engine type etc , y= car price which is output data.
                            
    ~ dependent data (y) [LABEL] : value which we predict.

    x and y both are available then supervised 

2.Unsupervised learning : ml which as unlabeled data.

     ~ x is available .

   ~ y is not available . 
  
    ~ ex: market basket analysis : prediciton of customer group based of income and expense.
